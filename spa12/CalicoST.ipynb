# 二、CalicoST 软件介绍

**CalicoST** 是一款专门针对**空间转录组 ($\text{SRT}$) 数据**开发的概率模型和算法，主要用于**癌症研究**。它的核心目标是同时推断肿瘤的**等位基因特异性拷贝数异常 ($\text{CNA}$)**，并重建肿瘤的**遗传进化和空间传播模式**（即系统发育地理）。

![](http://biomamba.com:34080/free.download/file.back/biomamba/chenao/calicost.png)

## 1. 核心功能与分析目标

| 目标领域 | 核心功能 | 生物学意义 |
| :--- | :--- | :--- |
| **基因组异常推断** | **等位基因特异性 $\text{CNA}$ 推断** | 识别传统总拷贝数分析中不可见的变异，如**拷贝中性杂合性缺失 ($\text{CNLOH}$)** 和**镜像亚克隆 $\text{CNA}$**。 |
| **肿瘤克隆分析** | **克隆标记与正常细胞混合处理** | 为每个 $\text{Spot}$ 分配癌症克隆标签，并能有效处理 $\text{SRT}$ 数据中常见的**正常细胞混合**问题，从而提高克隆鉴定的准确性。 |
| **进化与空间传播** | **系统发育地理 ($\text{Phylogeography}$) 重建** | 推断癌症克隆之间的**遗传进化关系**，并将这种进化与克隆在组织空间中的**传播**相结合（尤其适用于分析多张相邻切片数据）。 |

## 2. 关键优势

$\text{CalicoST}$ 通过整合基因表达和等位基因特异性信息，能够提供比现有方法**更高的 $\text{CNA}$ 准确性**，为研究肿瘤异质性、亚克隆进化及其空间格局提供了强大的工具。

然后，我们在复制的 `configuration_cna` 文件中修改以下路径：

- `spaceranger_dir` 为空间转录组输出文件目录  xx/outs
- `snp_dir` 为SNP信息文件目录（需要参考文件）
- `output_dir` 输出文件，*CalicoST* 用于写入推断克隆和 CNA 的输出目录

`configuration_cna` 文件中其他参数的默认值，如果需要调整其他参数信息


geneticmap_file: str
- 参考基因组文件

hgtable_file: str
- 基因组中基因位置的路径。这个文件应该是一个标签分隔的文件，包含以下列：gene_name、chrome、cdsStart、cdsEnd

normalidx_file: str, optional
- 存放的是正常（normal）spot 的索引，也就是在空间转录组数据里，被认为是正常组织（非肿瘤/非异常）的 spot 编号

tumorprop_file: str, optional
- 存放的是肿瘤（tumor）spot 的索引，也就是在空间转录组数据里，被认为是肿瘤组织（肿瘤/异常）的 spot 编号

filtergenelist_file: str, optional
- 基于先前知识的 CNA 推断中要排除的基因列表文件

filterregion_file: str, optional
- 要从 BED 格式的 CNA 推断中排除的基因组区域列表的文件。例如，HLA 区域。

logphase_shift: float, optional
- 调整相变中马尔科夫模型自转换的强度。值越高，自转换概率越高。默认值为 - 2.0

secondary_min_umi: int, optional
- 在基因组分割步骤中，一个基因组片段在假体斑点中的最小 UMI 计数。默认值为 300

n_clones: int
- 仅使用 BAF 信号推断的克隆数。默认值为 3。

n_clones_rdr: int, optional

- 使用 RDR 和 BAF 信号对每个 BAF 识别的克隆进行优化的克隆数。默认值为 2。

min_spots_per_clone: int, optional
- 调用克隆所需的最小位数应该是 100。默认值为 100。

min_avgumi_per_clone: int, optional
- 克隆所需的最小平均 UMI 计数。默认值为 10。

maxspots_pooling: int, optional
- 如果每个位点的 UMI 计数过低，CalicoST 将汇总这个相邻位点的数量，以推断每个 HMRF 步骤的克隆分配。默认值为 7。

nodepotential: str, optional
- 以下两个选项之一：“max” 或 “weighted_sum”。“max” 是指使用 HMM 的 MLE 解码来评估斑点存在于每个克隆中的概率。“weighted_sum” 是指使用完整的 HMM 后验概率来评估斑点存在于每个克罗中的概率。默认为 “weighted_sum”。

spatial_weight: float, optional
- HMRF 中的空间一致性强度。值越高，空间一致性越强。默认值为 1.0。

construct_adjacency_method: str, optional
- 从两种构建 HMRF 邻域图的方法中选择一种，即 “六边形” 或 “KNN”。“六边形” 假设spot定位形成六边形网格，如 Visium 平台所示。“KNN” 假设spot定位是任意的，并使用 K - 最近邻构建邻域图。默认为 “六边形”。

construct_adjacency_w: float, optional
- 如果使用 KNN 来构造邻接矩阵，CalicoST 允许将空间相似性与邻接矩阵的表达式相似性相结合。这个权重在 0 到 1 之间，指定了空间相似性的权重。默认值为 1.0。

n_states: int
- 等位基因特异性拷贝数在 CNA 推理的 HMM 中有记录。

t: float, optional
- HMM 的自转移概率。值越高，相邻基因组片段处于相同 CNA 状态的概率就越高。默认值为 1-1e-5。


max_iter: int, optional
- 在 HMM 中执行的 Baum-Welch 步骤数。默认值为 30。

tol: float, optional
- 终止 Baum-Welch 步骤的收敛阈值。默认值为 1e-4。

np_threshold: float, optional
- Neyman Pearson 统计量决定两个克隆的阈值具有不同的 CNA 事件。值越高，两个克隆就越容易合并。默认值为 1.0。

np_eventminlen: int, optional
- 被视为 CN 事件的连续基因组片段的最小数量。默认值为 10。
