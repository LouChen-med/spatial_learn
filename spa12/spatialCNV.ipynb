## <a id='toc4_4_'></a>[基因注释](#toc0_)

CNV分析需要明确每个基因所在的染色体位置，我们使用biomart数据库去注释

```python
from pybiomart import Server
import pandas as pd
```

```python
# 连接 Ensembl
server = Server(host='http://www.ensembl.org')
dataset = server.marts['ENSEMBL_MART_ENSEMBL'].datasets['hsapiens_gene_ensembl']
```

```python
# 获取人类基因的信息
gene_info = dataset.query(attributes=[
    'external_gene_name',   # gene symbol
    'ensembl_gene_id',      # ENSG
    'chromosome_name',      # 染色体
    'start_position',       # start
    'end_position'          # end
])
```

```python
gene_info
```

```python
# 将获取基因信息列表进行改造，符合软件所需格式
gene_info = gene_info.set_index("Gene name")
gene_info = gene_info.rename(columns={
    "Gene stable ID": "ensg",
    "Chromosome/scaffold name": "chromosome",
    "Gene start (bp)": "start",
    "Gene end (bp)": "end"
})
gene_info = gene_info[~gene_info.index.duplicated(keep='first')]
gene_info = gene_info.dropna(subset=['ensg','chromosome','start','end'])
```

```python
# 与空间转录组基因取交集
common_genes = adata.var_names.intersection(gene_info.index)
```

```python
# 对齐索引
adata.var.loc[common_genes, 'ensg'] = gene_info.loc[common_genes, 'ensg']
adata.var.loc[common_genes, 'chromosome'] = gene_info.loc[common_genes, 'chromosome']
adata.var.loc[common_genes, 'start'] = gene_info.loc[common_genes, 'start']
adata.var.loc[common_genes, 'end'] = gene_info.loc[common_genes, 'end']
```

```python
adata.var.loc[:, ['ensg','chromosome','start','end']].head()
```

```python
# 将 chromosome 转为字符串
adata.var['chromosome'] = adata.var['chromosome'].astype(str)

# 给所有非空值加上 "chr"
adata.var['chromosome'] = adata.var['chromosome'].apply(lambda x: f"chr{x}" if x != 'nan' else x)

# 检查结果
adata.var['chromosome'].unique()[:20]
```

```python
# 这里删除非常规染色体（保留1-22常规染色体）
import re
# 1️⃣ 筛选所有 ensf/chromosome/start/end 非 NA 的基因
mask = adata.var[['ensg','chromosome','start','end']].notna().all(axis=1)
mask

# 2️⃣ 保留常规染色体
# 假设 chromosome 已经是 'chr1', 'chr2', ..., 'chrX', 'chrY', 'chrMT'
chrom_mask = adata.var['chromosome'].str.match(r'^chr([1-9]|1[0-9]|2[0-2]|X|Y|MT)$')
chrom_mask.value_counts()

# 3️⃣ 同时满足两条件
final_mask = mask & chrom_mask

# 4️⃣ 筛选基因
adata = adata[:, final_mask].copy()

# 5️⃣ 检查
adata.var.loc[:, ['ensg','chromosome','start','end']].head()
```

```python
# 保存最终用于分析的adata对象
adata.write('/home/chenao/jupyter_home/downstream/process/infer_CNV/inferCNV/inferCNV.data')
```
