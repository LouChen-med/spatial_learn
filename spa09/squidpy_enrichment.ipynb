# 二、邻域分析

## 1. 空间转录组图像特征提取：结合形态与分子信息

### 核心目标

在 $\text{Visium}$ 空间转录组分析中，$\text{Spot}$ 的基因表达数据提供了分子信息，而高分辨率组织图像则提供了**细胞形态和组织结构**的视觉信息。

使用 $\text{Squidpy}$ 库的 `squidpy.im.calculate_image_features()` 函数，我们可以从组织图像中计算每个 $\text{Visium}$ 斑点（$\text{Spot}$）的**图像特征**。这些特征将作为一个 $\text{obs}$ $\times$ $\text{特征}$ 矩阵，与 $\text{obs}$ $\times$ $\text{基因表达}$ 矩阵一起存储在 $\text{AnnData}$ 对象中，用于下游的联合分析。

## 2. 图像特征的价值：互补与融合

通过提取图像特征，我们旨在获得与基因表达值**相似且互补**的信息：

* **相似信息**：例如，在具有两种形态不同细胞类型的组织中，图像特征可以捕获细胞的**形态差异**。这种细胞类型信息随后同时包含在基因表达值和组织图像特征中。
* **互补信息**：图像特征能够量化**组织纹理**、**细胞密度**、**结构复杂度**等，这些信息不能完全从基因表达中获得。将图像特征与基因表达结合，可以更全面地定义和识别组织区域。

## 3. 应用

提取多尺度和多维度特征：

在这里，我们将提取不同尺寸（$\text{Crop}$ $\text{Size}$）和尺度（$\text{Scale}$）的总结特征，以便计算多尺度特征和分割特征。

| 概念 | 描述 | 目的 |
| :--- | :--- | :--- |
| **尺寸** | 围绕每个 $\text{Spot}$ 裁剪的局部图像区域大小。 | 决定特征捕获的局部范围。 |
| **尺度 ($\text{Scale}$)** | 在不同的图像分辨率下计算特征。 | 捕获不同大小的结构信息（如单个细胞 vs. 细胞群落）。 |
| **总结特征** | 图像的统计特性（如平均亮度、熵、纹理）。 | 量化 $\text{Spot}$ 周围组织的通用视觉特性。 |
| **分割特征** | 基于细胞或核分割结果提取的形态特征。 | 细化到细胞形状、大小和密度等信息。 |

通过这种方法，我们能够将高分辨率组织形态信息有效地转换为定量特征，实现与分子数据的深度融合分析。

## 3. 计算不同尺度的特征

```python
# --- 循环计算不同尺度的图像摘要特征 ---

for scale in [1.0, 2.0]:
    # 迭代计算两个不同的尺度 (scale=1.0 和 scale=2.0) 的特征。
    # 目的：捕获组织结构在不同分辨率或大小上的信息，实现多尺度分析

    feature_name = f"features_summary_scale{scale}"
    # 构造特征结果存储的键名 (key_added)，例如 "features_summary_scale1.0"
    sq.im.calculate_image_features(
        adata,
        # AnnData 对象：用于匹配 Spot 坐标和存储结果
        img.compute(),
        # 图像数据：ImageContainer 对象，通过 .compute() 获取图像数据进行处理
        features="summary",
        # 特征类型：指定计算摘要特征 (Summary Features)，如颜色均值、标准差、熵等  
        library_id = 'square_016um',
        # 库 ID：指定当前 AnnData 对象中用于识别 Spot 坐标的库名称 
        key_added=feature_name,
        # 结果存储键：指定将计算得到的特征矩阵存储在 adata.obsm 中的键名  
        scale=scale,
        # 尺度：指定当前迭代使用的图像缩放尺度
    )
    # 作用：计算每个 Visium Spot 局部区域的图像统计特征，并将结果添加到 adata.obsm 中
```

## 4. 数据聚合

```python
# --- 聚合图像特征到单个 DataFrame ---
adata.obsm["features"] = pd.concat(
    [adata.obsm[f] for f in adata.obsm.keys() if "features_summary" in f],
    # 列表推导式：从 adata.obsm 中筛选出所有名称中包含 "features_summary" 的键（即前面计算的两个尺度特征）
    # 然后将这些特征矩阵提取出来
    axis="columns",
    # 拼接方式：沿着列 (axis="columns") 拼接这些特征矩阵
    # 目的：创建一个新的 DataFrame，其中行是 Spot，列是所有尺度和类型的图像特征
)
# 作用：将多尺度计算的图像特征矩阵合并为一个大的特征矩阵，并存储在 adata.obsm["features"] 中

# --- 确保特征名称唯一性 ---
adata.obsm["features"].columns = ad.utils.make_index_unique(
    adata.obsm["features"].columns
)
# 作用：调用 anndata.utils.make_index_unique 函数，处理合并后的特征矩阵的列名（特征名）
# 目的：避免由于不同尺度的特征具有相同的名称而导致的列名重复问题，确保每个特征都具有唯一的标识符
```

## 5. 数据基础分析

```python
# 写一个快速基础分析的函数

# 基于图像特征进行聚类
def cluster_features(features: pd.DataFrame, like=None) -> pd.Series:
    """
    Calculate leiden clustering of features.
    Specify filter of features using `like`.
    """
    # ----------------------------------------------------
    # 1. 过滤特征
    # ----------------------------------------------------
    # filter features
    if like is not None:
        # 如果指定了 'like' 关键字（例如 'summary' 或 'scale1.0'），则仅保留名称中包含该字符串的特征列
        features = features.filter(like=like)
    # ----------------------------------------------------
    # 2. 准备 AnnData 对象和预处理
    # ----------------------------------------------------
    # create temporary adata to calculate the clustering
    adata = ad.AnnData(features)
    # 将特征 DataFrame 转换为临时的 AnnData 对象，以便使用 Scanpy 的工具链
    
    # important - feature values are not scaled, so need to scale them before PCA
    sc.pp.scale(adata)
    # 标准化：对图像特征进行 Z-score 标准化（均值=0, 方差=1）
    # 目的：防止特征值范围大的特征主导 PCA 结果
    
    # ----------------------------------------------------
    # 3. 降维和聚类
    # ----------------------------------------------------
    # calculate leiden clustering
    sc.pp.pca(adata, n_comps=min(10, features.shape[1] - 1))
    # PCA 降维：计算主成分
    # - n_comps：取 10 和特征数减 1 中的较小值作为主成分数量，避免特征太少时出错
    
    sc.pp.neighbors(adata)
    # 构建 KNN 邻域图：基于 PCA 降维结果，计算 Spot 之间的相似性
    
    sc.tl.leiden(adata, resolution=0.3)
    # Leiden 聚类：基于 KNN 图执行聚类
    # - resolution=0.3：设定较低的分辨率，倾向于生成较少、较大的群集（组织区域）

    # ----------------------------------------------------
    # 4. 返回结果
    # ----------------------------------------------------
    return adata.obs["leiden"]
    # 返回聚类结果：一个 Pandas Series，其中索引是 Spot 名称，值是 Spot 所属的 Leiden 簇 ID
```

## 6. 可视化聚类信息

```python
# 计算特征聚类
adata.obs["features_cluster"] = cluster_features(
    adata.obsm["features"], 
    # 输入特征：使用 adata.obsm 中包含所有尺度图像特征的 DataFrame。
    like="summary"
    # 过滤条件：指定只使用特征名称中包含 "summary" 的列。
    # 目的：确保聚类仅基于图像的摘要特征（如颜色、亮度、纹理统计），排除了其他可能的分割特征（如果存在）
)
# 作用：
# 1. 对筛选出的图像摘要特征进行 Z-score 标准化
# 2. 运行 PCA 降维
# 3. 构建 KNN 邻域图
# 4. 执行 Leiden 聚类（分辨率 R=0.3）
# 5. 将聚类结果（簇 ID）存储在 adata.obs 的新列 "features_cluster" 中

# 结果："features_cluster" 列代表了基于 Spot 局部形态和组织纹理相似性划分出的空间区域
```

## 7. 邻域富集

### 7.1 空间邻域富集分析：识别组织中 Spot 簇的共现模式

**核心概念**

邻域富集分析 (`Neighborhood Enrichment Analysis`) 旨在识别在组织空间中**倾向于共同邻近**或**倾向于相互排斥**的 Spot 簇（例如，基于基因表达、图像特征或细胞类型划分的区域）。

简而言之，这是一个关于簇在空间上邻近程度的**富集得分**：

* **高得分 (富集)**：如果属于**两个不同簇**的 Spot 经常彼此接近，那么它们将获得高得分，可以定义为正在被**富集**。
* **低得分 (耗尽)**：如果它们距离很远，因此很少成为邻域，那么得分将很低，可以定义为已**耗尽**。

### 7.2 分析步骤

由于该富集分析函数作用于连接矩阵，我们需要依次计算空间邻域矩阵，然后计算富集得分。

#### 计算空间邻域矩阵

我们需要构建一个连接矩阵，描述 $\text{Spot}$ 之间的空间邻近关系。

**使用函数：** `squidpy.gr.spatial_neighbors()`

* **目的**：根据 $\text{AnnData}$ 对象中存储的**物理空间坐标**，计算每个 $\text{Spot}$ 的空间邻居，生成一个邻域图（连接矩阵）。

#### 计算邻域富集得分

**使用函数：** `squidpy.gr.nhood_enrichment()`

该函数计算每对簇在空间上邻近程度的富集得分。

* **得分计算**：基于观察到的簇间连接数与随机预期连接数的比较。
* **统计检验**：富集得分基于**排列 ($\text{Permutation}$ $\text{Test}$)** 计算。
* **参数**：可以使用 `n_perms` 参数设置排列次数（默认为 1000）。

最终结果通常以热图形式展示，**清晰地揭示每对组织区域或细胞簇之间的空间相互作用模式**。

```python
sq.gr.spatial_neighbors(adata)
sq.gr.nhood_enrichment(adata, cluster_key="celltype")
sq.pl.nhood_enrichment(adata, cluster_key="celltype")
```

### 7.3 配体 - 受体相互作用分析

继续分析，展示几种与空间分子数据分析非常相关的特征级方法。例如，在定量化簇共现之后，我们可能会对寻找可能驱动细胞通信的分子实例感兴趣。这自然转化为配体 - 受体相互作用分析。在 Squidpy 中，我们提供了一个快速重新实现流行方法 CellPhoneDB [Efremova 等，2020](代码 < https://github.com/teichlab/cellphonedb>_), 并使用流行的 Omnipath 数据库 [Türei 等，2016] 扩展了其注释配体 - 受体相互作用对的数据库。你可以使用 squidpy.gr.ligrec () 对所有簇对和所有基因 (在几秒钟内，不离开这个笔记本) 运行分析。此外，我们将直接可视化结果，过滤低表达基因 (使用 means_range 参数), 并增加调整后 p 值的阈值 (使用 alpha 参数)。我们还将仅对一个源组 (海马体簇) 和两个目标组 (Pyramidal_layer_dentate_gyrus 和 Pyramidal_layer_簇) 进行可视化子集。

```python
sq.gr.ligrec(
    adata2,
    # 输入的 AnnData 对象：包含基因表达数据和聚类结果
    n_perms=100,
    # 排列次数：指定用于计算 p 值的排列检验次数为 100 次
    # 目的：通过与随机背景进行比较，确定观察到的配体-受体对活性的统计显著性。
    # 提示：实际分析中，通常建议使用更高的排列次数（如 1000 次）以获得更稳健的 p 值
    cluster_key="cluster",
    # 聚类键：指定 AnnData.obs 中用于定义相互作用 Spot 组（发送方和接收方）的列名
    # 目的：分析将针对 "cluster" 列中的每一对簇进行，推断它们之间的通讯强度
)
# 作用：
# 1. 自动查询内置的配体-受体数据库（例如 CellTalkDB 或 Omnipath）。
# 2. 计算每个配体-受体对在不同 Spot 簇对之间的表达富集程度。
# 3. 使用排列检验计算富集得分和相应的统计显著性（p 值/FDR）。

# 结果：
# 分析结果（包括得分和 p 值）将存储在 adata2.uns 中，通常键名为 'ligrec_{cluster_key}'，
# 可用于后续绘图和识别组织区域之间的关键通讯轴
```

```python
sq.pl.ligrec(
    adata2,
    cluster_key="cluster",
    # 聚类键：指定用于区分发送方和接收方 Spot 组的列名
    source_groups="Hippocampus",
    # 发送方组（Source Groups）：指定作为配体发送方的细胞簇为 "Hippocampus"（海马区）
    target_groups=["Pyramidal_layer", "Pyramidal_layer_dentate_gyrus"],
    # 接收方组（Target Groups）：指定作为受体接收方的细胞簇列表，这里是两个锥体细胞层
    means_range=(1, np.inf),
    # 均值范围：过滤相互作用的强度范围。这里要求配体和受体的平均表达值必须大于 1
    # 目的：排除表达量极低的配体-受体对
    alpha=1e-4,
    # 统计显著性阈值：仅显示 p 值（或 FDR）小于 1e-4 的配体-受体
    # 目的：确保只展示统计学上高度显著的通讯
    swap_axes=True,
    # 轴交换：将热图的 X 轴和 Y 轴进行交换。
    # 目的：可能为了优化图表的布局或使结果更易于阅读（例如，将配体放在 Y 轴，接收簇放在 X 轴）。
)
# 作用：
# 生成一个热图，可视化 "Hippocampus" 簇发送给 "Pyramidal_layer" 和 "Pyramidal_layer_dentate_gyrus" 簇之间，
# 那些在统计上显著且表达强度高的配体-受体对。热图的颜色或大小通常代表通讯强度或 p 值
```

点状图可视化提供了一组有趣的候选配体 - 受体注释，可能涉及海马体中的细胞相互作用。例如，更精细的分析将会将这些结果与反褶皱方法的结果进行整合，以了解组织这个区域中单细胞细胞类型的比例
